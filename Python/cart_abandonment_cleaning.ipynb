```python
# This script checks if a specified folder exists and lists all CSV files in that folder.
from pathlib import Path

data_folder = Path(r"C:\Users\DELL\Desktop\Data Analytics\Project\Dataset")

print("Folder exists:", data_folder.exists())
print("CSV files:", [file.name for file in data_folder.glob("*.csv")])
```

```python
# List all CSV files with their sizes
for file in sorted(data_folder.glob("*.csv")):
    size_mb = file.stat().st_size / (1024 * 1024)
    print(file.name, f"{size_mb: .1f} MB)")
```

```python
# List all CSV files with their column names
import pandas as pd
for file in sorted(data_folder.glob("*.csv")):
    columns = pd.read_csv(file, nrows=0).columns.tolist()
    print(file.name, columns)
```

```python
# Count the number of rows in each CSV file
chunk_size = 500_000

for file in sorted(data_folder.glob("*.csv")):
    row_count = 0

    for chunk in pd.read_csv(file, chunksize=chunk_size):
        row_count += len(chunk)

    print(file.name, f"{row_count:,} rows")
```

```python
# Count the number of missing values in each CSV file
missing_totals = {}

for file in sorted(data_folder.glob("*.csv")):
    file_missing = {}

    for chunk in pd.read_csv(file, chunksize=chunk_size):
        missing_in_chunk = chunk.isna().sum()

        for column, count in missing_in_chunk.items():
            file_missing[column] = file_missing.get(column, 0) + count

    missing_totals[file.name] = file_missing
```

```python
# Print the missing values for each file
for file_name, counts in missing_totals.items():
    print(f"\n{file_name}")
    for column, count in counts.items():
        print(f"{column}: {count:,}")
```

```python
# Count the number of occurrences of each event type in each CSV file
event_totals = {}

for file in sorted(data_folder.glob("*.csv")):
    file_events = {}

    for chunk in pd.read_csv(
        file,
        usecols=["event_type"],
        chunksize=chunk_size
    ):
        event_counts = chunk["event_type"].value_counts()

        for event, count in event_counts.items():
            file_events[event] = file_events.get(event, 0) + count

    event_totals[file.name] = file_events

for file_name, counts in event_totals.items():
    print(f"\n{file_name}")
    for event, count in counts.items():
        print(f"  {event}: {count:,}")
```

```python
october_output = cleaned_folder / "2019-Oct_cleaned.csv"

first_chunk = True
rows_written = 0

for chunk in pd.read_csv(october_file, chunksize=500_000):
    chunk["event_time"] = pd.to_datetime(chunk["event_time"], utc=True)

    chunk["event_type"] = chunk["event_type"].str.strip()
    chunk["category_code"] = chunk["category_code"].str.strip()
    chunk["brand"] = chunk["brand"].str.strip()
    chunk["user_session"] = chunk["user_session"].str.strip()

    chunk.to_csv(
        october_output,
        mode="w" if first_chunk else "a",
        index=False,
        header=first_chunk
    )

    rows_written += len(chunk)
    print(f"Processed {rows_written:,} rows")

    first_chunk = False

print("Saved cleaned file:", october_output)
```
